## 字符串的显式构造
Python 2和Python 3中, 使用一对**半角**双引号(或一对**半角**单引号)将连续的若干个字符括注, 所声明的对象是字符串(`str`)对象. 

将字符串内容与字符串以外其他部分之间分隔的双引号(或单引号)对, 称为字符串定界符. Python解释器从代码的首行开始, 检测到字符串定界符交替出现时, 会将相邻定界符之间的内容交替解释为字符串或者其他代码
* 当字符串内使用转义规则表示半角单(双)引号时, 转义的单(双)引号仍被解释为字符串内容的一部分, 而非定界符. 

字符串被显式声明后, 将根据其中字符的排列顺序, 将其中的字符逐个按照特定的编码规则 (可以视为字符集到自然数集$\mathbb{N}$的映射) 转化为无重叠无间隔, 在RAM上连续记录的整数序列. 

* 字符串的默认编码规则与解释器设置有关, 可通过`sys`模块的`getdefaultenocoding()`获得; 
* 在Python 3中, 解释器初始化后采用的默认字符编码为`UTF-8`; 
* 在Python 2中, 解释器初始化后采用的默认字符编码与所在的操作系统有关. 
* 如需在Python 2中强制声明采用`UTF-8`编码的字符串, 需在字符串的定界符之前前插入`u`. 例如: 
```python
u"ABC"
```
表示采用`UTF-8`编码的字符串`"ABC"`, 与解释器采用的默认编码方式无关.

## 字符串的隐式构造
字符串的隐式构造方式包括: 
* 调用对象`__str__`或`__repr__`方法; 
* 字符串拼接; 
* 构造格式化字符串. 

### 利用其他对象构造字符串
|目标|用法|其他等效形式|
|:-:|:-|:-|
|调用`__str__`方法|`obj.__str__()`|`str(obj)`|
|调用`__repr__`方法|`obj.__repr__()`|`repr(obj)`|

调用对象`obj`的`__str__`或`__repr__`方法, 利用对象实例中的属性信息构造字符串, 用于表示`obj`的主要内容和特性. 
* `print(obj)`在将`obj`的文本形式输出到指定的输出流之前, 将自动调用`obj.__str__()`. 
* 在python解释器命令行中直接键入`obj`查看对象状态, 或者在`pdb`子会话中使用`p obj`查看对象状态, 输出到指定输出流时, 自动调用的是`obj.__repr__()`. 
* 在`jupyter notebook`输入单元中直接键入`obj`查看对象状态, 若对象支持其他特殊方法 (如`_repr_latex_()`或`_repr_svg_()`), 将优先调用特定方法, 并根据相应方法返回的文本, 以及对象的类型, 以特定的格式渲染至笔记本文档的输出单元, 当`obj`不支持上述方法时, 才会调用`__repr__()`

````python
#比较numpy.polynomial.Polynomial不同文本输出方法的文本差异. 
from numpy import polynomial as nppn; 
quad = nppn.Polynomial((1, -2, 1)); 
#在jupyter notebook的输入单元直接键入quad会显示其公式, 过程中
#  调用其_repr_latex_方法得到LaTeX代码并渲染至输出单元; 
#在python解释器命令行会话下直接键入quad会输出quad.__repr__()
print(quad.__str__(), quad.__repr__(), quad._repr_latex_(), 
    sep="\n"); 
````

### 字符串的格式化构造
本文档仅讨论基于`str.format(*args, **kwargs)`方法的字符串格式化构造. 
> `python`支持三种字符串格式化语法: `__mod__`, `format`和`f-string`: 
> 
> |语法|引入版本|特征|
> |:-|:-:|:-|
> |`str.__mod__`<br>(`'%xxx' % obj`)|`1.x`|语法风格与C语言类似, `%`的数量和位置<br>必须与传入的对象**一一对应**, **容错率较低**|
> |`str.format`方法|`2.6`|可自定义参数在格式化后嵌入的位置; <br>可**兼容Python 2.7和Python 3.x**|
> |`f-string`|`3.6`|**兼容性较差**, 部分发行年代较早的二次开发<br>平台使用的Python解释器不支持该语法|

用法: 
```python
str_template.format(*args, **kwargs)
```
* `str_template` 格式化显示时使用的模板文本. 
* `args`, `kwargs` 以位置参数和关键字参数指定的, 需要在格式化后嵌入模板文本的内容. 其中`args`的内容将在传入`format`方法时被解包为`tuple`, `kwargs`的内容将被解包为`dict`, 键值对关系即为关键字与`=`右侧传入值的关系. 

要将`args`和`kwargs`的内容格式化并嵌入`str_template`, 需要在嵌入的位置插入一对左右花括号`{}`, 并在花括号内指定要嵌入的参数及其显示形式. 

`{...}`定界符内的内容包含三部分: 

`{` `idx` `preproc` `display` `}`

* `idx` 需要嵌入的参数索引, 接受一个`int`或者合法的`python`关键字标识符. 
    * 当使用`int`时, 接受的参数是`args`解包的`tuple`中以该`int`为下标的对象`*args[int(idx)]`; 
    * 当使用关键字标识符时, 接受的参数是`kwargs`解包的`dict`中, 以该标识符为键时对应的值`**kwargs[idx]`; 
    * 可以使用下标访问`idx`所指的对象的成员, 也可以使用`.`访问其属性, 但是不可调用其方法
* `preproc` 在嵌入前对参数的预处理
|选项|作用|
|:-|:-|
|(省略)|不作处理|
|`!s`|调用参数的`__str__`方法|
|`!r`|调用参数的`__repr__`方法|
* `display` 在完成(或跳过)预处理后对嵌入后子串显示格式的控制
    * `display`可以省略, 当`display`和`preproc`同时省略时, 将对嵌入的参数直接调用`__str__`方法; 
    * `display`不予省略时, 必须以`:`开头; 
    * `display`分为多个部分: 对齐方式, 长度与精度, 格式化类型
|部分|格式|备注|
|:-:|:-|:-|
|对齐与填充|`p>`, `p<`或`p^`|`p>` 左对齐<br>`p<` 右对齐<br>`p^` 居中对齐<br>`p`为对齐后在两端需要填充的字符|
|长度与精度|`l`或`l.pr`|`l` 格式化后的子串最小长度<br>当格式化后的实际长度小于`l`时, <br>将采用对齐和填充规则; <br>`pr` 当格式化类型为`f`, `%`, `e`或`E`时, <br>指定**格式化后的**小数部分需要保留的位数<br>修约方式采用"四舍五入"规则, 如需采用<br>其他规则, 需构造`Decimal`对象|
|格式化类型|`c`, `s`, <br>`d`, `b`, `o`, `x`, <br>`f`, `%`, `e`, `E`|`c` 将整型对象按照对应的编码规则<br>转换为字符; <br>`s` 对字符串对象不进行转换; <br>`d`, `b`, `o`, `x` 分别以十进制, 二进制, <br>八进制, 十六进制显示整型对象; <br>`f`, `%`, `e`, `E` 分别以小数, 百分数, <br>科学计数法(小写`e`), 科学计数法<br>(大写`E`)形式显示浮点型对象. |

````python
#先嵌入第二个位置参数, 再嵌入以optn为关键字的参数, 最后嵌入第一个位置参数
"{1}{optn}{0}".format("A", "B", optn="C")
````

## 字符串格式构造的安全隐患: 命令注入攻击

### 概述
在`python`的部分库中, 调用函数执行自定义命令时, 命令及其参数将被置于同一字符串中作为函数的参数传入. 
例如: 使用`eval`, `exec`执行字符串形式的`python`源码, 使用`os.system`执行系统命令行, 使用`sqlite3.Connection.execute`在打开的数据库中执行`SQL`命令等. 

若开发者在功能模块里允许用户输入与命令有关的参数, 并通过格式化字符串的方式, 在**未经检查或预处理的情况下, 将参数直接嵌入**命令的字符串中, 
则用户可以利用上述缺陷, 通过**向参数中引入特殊字符和语法, 使得嵌入参数后的命令字符串被"注入"其他命令**, 用于执行预期以外的其他操作, 从而达到侵入系统, 窃取, 篡改和删除数据的目的. 
这一过程称为"命令注入攻击", 而能够被攻击者利用的缺陷称为"注入漏洞". 

### 注入攻击演示与分析

假设开发者提供一个程序, 通过二分法计算关于$x$的一元函数在特定区间内的零点. 

````python
import types; 

def bisection(a: float, b: float, f: types.FunctionType, eps: float): 
    x_0, x_1 = a, b; 
    y_0, y_1 = f(x_0), f(x_1); 
    if y_0 * y_1 > 0: 
        raise ValueError; 
    x_c = (x_0 + x_1) / 2; 
    while x_1 - x_0 > 2 * eps: 
        y_c = f(x_c); 
        if y_c == 0: 
            return x_c; 
        elif y_0 * y_c < 0: 
            x_1 = x_c; y_1 = y_c; 
        else: 
            x_0 = x_c; y_0 = y_c; 
        x_c = (x_0 + x_1) / 2; 
    return x_c; 
````

开发者将函数封装成可供用户调用的形式, 要求用户输入区间左, 右端点, 函数解析式, 以及零点的精度. 
由于表达式的输入由用户通过命令行或者图形界面中的文本框输入, 因此需要转化为python中的匿名函数. 

````python
def bisection_input(a: str, b: str, f: str, eps: str): 
    a_real = float(a); 
    b_real = float(b); 
    eps_real = float(eps); 
    f_func = eval("lambda x: {expr:s}".format(expr=f)); #这里有一个注入漏洞
    return bisection(a_real, b_real, f_func, eps_real); 
````

用户开始通过命令行调用这个函数. 
假设用户利用程序计算函数
$$f(x)=x^2-2$$
在区间
$$\lbrace x \in \mathbb{R} | 1 \le x \le 2\rbrace$$
上的零点$x_0 = \sqrt{2} \approx 1.41421 \dots$, 要求截断误差$\varepsilon \le 10^{-7}$. 

在开发者的预期内, 用户的输入可能是: 

````python
print(bisection_input("1", "2", "x ** 2 - 2", "1e-7"), file=sys.stdout);
````

然而用户可能借助参数`f`中代码被直接组合后`eval`的漏洞, 发起注入攻击: 
* 本示例注入的是无害命令: 向标准错误流`sys.stderr`打印解释器采用的浮点数标准. 

````python
print(bisection_input("1", "2", 
    "(x ** 2 - 2, print(__import__('sys').float_info, \
    file=__import__('sys').stderr, end='\\r'))[0]", 
    "1e-7"), file=sys.stdout);  
````

注入攻击的代码导致
```python
bisection_input(a, b, f, eps)
```
在传入`f`的实参后, 构造了
```python
lambda x: (x ** 2 - 2, 
    print(sys.float_info, file=sys.stderr, end='\r')
)[0]
```
需要返回`tuple`的首个对象(`x ** 2 - 2`), 但是`tuple`在访问下标前需要先对所有的元素都求值, 
因此当`bisection`每次调用从`bisection_input`传入的`f_func(x)`时, 会先计算
```python
x ** 2 - 2
```
(此处临时记为`y`), 再执行注入的命令
```python
print(sys.float_info, file=sys.stderr, end='\r')
```
(函数执行后的返回值为`None`), 对所得的二元`tuple`(即`(y, None)`)取首个元素`(...)[0]`, 即为`y`. 

因此, 从`bisection`函数的视角考察, 执行上述`f_func(x)`的效果与执行
```python
(lambda x: x ** 2 - 2)(x)
```
的结果是相同的, 不会影响后续程序的执行; 但从`bisection_input`的视角考察, 
由于其调用过程未对传入的`f`中的字符串内容进行检查, 被额外执行了注入的命令. 

在实际的生产环境中, 攻击者借由此类命令发现程序漏洞后, 会进一步尝试注入其他破坏性命令. 
因此开发者不能过度信任用户输入.
